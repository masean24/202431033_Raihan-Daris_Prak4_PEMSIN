```python
# Mengambil kolom numerikal
numerical_features = [feature for feature in data.columns if data[feature].dtypes != 'O']
print("Jumlah fitur numerik = ", 
      len(numerical_features))
print("Daftar fitur numerik = ", 
      numerical_features)
display(data[numerical_features].head())
# Raihan Daris Ramadhan 202431033
```

```python
#Mengambil kolom kategorikal
categorical_features = [feature for feature in data.columns if data[feature].dtypes == 'O']
print("Jumlah fitur kategorikal = ", 
      len(categorical_features))
print("Daftar fitur kategorikal = ", 
      categorical_features)
display(data[categorical_features].head())
# Raihan Daris Ramadhan 202431033
```

```python
# Menampilkan data unik pada fitur kategorikal
for i in categorical_features:
    print(data[i].unique())
    # Raihan Daris Ramadhan 202431033
```

```python
# Mengubah kolom kategorikal menjadi angka untuk klasifikasi

from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
for feature in categorical_features:
    print(data[feature].unique())
    data[feature] = le.fit_transform(data[feature])
    print(data[feature].unique())
    # Raihan Daris Ramadhan 202431033
```

```python
data.head()
# Raihan Daris Ramadhan 202431033
```

```python
# Mengambil kolom untuk fitur dan label
x = data.drop(['Drug'], axis=1)
y = data['Drug']
print(x)
print(y)
# Raihan Daris Ramadhan 202431033
```

```python
#Membagi data menjadii data test dan data train
from sklearn.model_selection import train_test_split as tts 

# Raihan Daris Ramadhan 202431033

x_train, x_test, y_train, y_test = tts(x,y, test_size = 0.25, random_state = 0)
```

```python
#Melatih model gaussian naive bayes
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
model.fit(x_train, y_train)
# Raihan Daris Ramadhan 202431033
```

```python
y_pred_train = model.predict(x_train)
y_pred_test = model.predict(x_test)
# Raihan Daris Ramadhan 202431033
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix

print("Evaluasi Model")
cm_train = confusion_matrix(y_train, y_pred_train)

# Cetak matriks dan classification report
print("Confusion Matrix (Train):")
print(cm_train)
print("\nClassification Report (Train):")
print(classification_report(y_train, y_pred_train))

# Visualisasi Confusion Matrix
plt.figure(figsize=(6, 4))
sns.heatmap(
    cm_train,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=np.unique(y_train),
    yticklabels=np.unique(y_train)
)
plt.title("Confusion Matrix - Training")
plt.xlabel("Label Terprediksi")
plt.ylabel("Label Aktual")
plt.tight_layout()
plt.show()
# Raihan Daris Ramadhan 202431033
```

```python
print("Evaluasi Model - Testing")
cm_test = confusion_matrix(y_test, y_pred_test)

print("Confusion Matrix (Test):")
print(cm_test)
print("\nClassification Report (Test):")
print(classification_report(y_test, y_pred_test))

plt.figure(figsize=(6, 4))
sns.heatmap(
    cm_test,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=np.unique(y),
    yticklabels=np.unique(y)
)
plt.title("Confusion Matrix - Testing")
plt.xlabel("Label Terprediksi")
plt.ylabel("Label Aktual")
plt.tight_layout()
plt.show()
# Raihan Daris Ramadhan 202431033
```
